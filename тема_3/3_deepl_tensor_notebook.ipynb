```python
# Broadcasting (трансляція): множимо тензор 2×4 на тензор 1×4
# PyTorch «розтягує» 1×4 до 2×4 і множить поелементно
rand = torch.rand(2, 4)
doubled = rand * (torch.ones(1, 4) * 2)

print(rand)     # Оригінальний тензор
print(doubled)  # Отримуємо подвоєні значення (кожен елемент × 2)
```

```python
# Приклади broadcasting, що ПРАЦЮЮТЬ:
a = torch.ones(4, 3, 2)

# (4,3,2) × (3,2): вимір 1 відсутній → розтягується на 4 шари
b = a * torch.rand(3, 2)
print(b)

# (4,3,2) × (3,1): останній вимір=1 → розтягується на 2 стовпці
c = a * torch.rand(3, 1)
print(c)

# (4,3,2) × (1,2): середній вимір=1 → розтягується на 3 рядки
d = a * torch.rand(1, 2)
print(d)
```

# Logistic regression with PyTorch

```python
# Підключаємо бібліотеки для логістичної регресії на PyTorch
import numpy as np                              # Числові обчислення
import pandas as pd                             # Робота з таблицями

from sklearn.impute import SimpleImputer        # Заповнення пропусків
from sklearn.model_selection import train_test_split  # Розбиття train/test
from sklearn.metrics import accuracy_score      # Вимірювання точності

import torch                                    # PyTorch: тензори
import torch.nn as nn                           # Шари, активації, loss-функції
import torch.optim as optim                     # Оптимізатори (SGD, Adam)

import warnings
warnings.filterwarnings('ignore')               # Ховаємо попередження
```

## Data Procesing

```python
# Завантажуємо датасет Spaceship Titanic
# Для Colab: замініть шлях на URL:
# df = pd.read_csv('https://raw.githubusercontent.com/goitacademy/DEEP-LEARNING-FOR-COMPUTER-VISION-AND-NLP/main/data/Module_1_Lecture_2_Class_Spaceship_Titanic.csv')
df = pd.read_csv('../data/Module_1_Lecture_2_Class_Spaceship_Titanic.csv')
df = df.set_index('PassengerId')  # Робимо ID пасажира індексом

TARGET = 'Transported'  # Що передбачаємо: чи був пасажир транспортований (0/1)
FEATURES = [col for col in df.columns if col != TARGET]  # Усі ознаки крім цілі
```

```python
# --- Заповнюємо пропуски числових колонок медіаною ---
imputer_cols = ["Age", "FoodCourt", "ShoppingMall", "Spa", "VRDeck", "RoomService"]
imputer = SimpleImputer(strategy='median')  # Стратегія: медіана (стійка до викидів)
imputer.fit(df[imputer_cols])               # Вивчаємо медіани
df[imputer_cols] = imputer.transform(df[imputer_cols])  # Заповнюємо пропуски

# --- Заповнюємо пропуски категоріальних змінних фіктивними значеннями ---
df["HomePlanet"].fillna('Gallifrey', inplace=True)
df["Destination"].fillna('Skaro', inplace=True)

# --- Додаємо індикатори пропусків (1 = було пропущено) ---
df['CryoSleep_is_missing'] = df['CryoSleep'].isna().astype(int)
df['VIP_is_missing'] = df['VIP'].isna().astype(int)

# --- Заповнюємо логічні змінні та перетворюємо True/False → 1/0 ---
df["CryoSleep"].fillna(False, inplace=True)
df["VIP"].fillna(False, inplace=True)
df["CryoSleep"] = df["CryoSleep"].astype(int)
df["VIP"] = df["VIP"].astype(int)

# --- One-Hot Encoding: перетворюємо категорії (текст) → числа ---
dummies = pd.get_dummies(df.loc[:, ['HomePlanet', 'Destination']], dtype=int)
df = pd.concat([df, dummies], axis=1)
df.drop(columns=['HomePlanet', 'Destination'], inplace=True)

# --- Перетворюємо ціль у числа, видаляємо текстові колонки ---
df[TARGET] = df[TARGET].astype(int)
df.drop(["Name", "Cabin"], axis=1, inplace=True)
```

```python
# Розбиваємо дані на тренувальні (67%) та тестові (33%)
X = df.drop(TARGET, axis=1).values   # Вхідні ознаки → NumPy масив
y = df[TARGET].values                 # Цільова змінна → NumPy масив

X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, test_size=0.33, stratify=y
)
# Отримуємо 4 масиви: X_train, X_test, y_train, y_test
```

```python
# Перетворюємо NumPy масиви → тензори PyTorch (float32 для нейромереж)
X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.float32)

X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.float32)
# Отримуємо тензори, готові до подачі в модель
```

## Model

```python
# Визначаємо модель логістичної регресії
# Наслідуємо від nn.Module — базового класу всіх моделей PyTorch

class LogisticRegression(nn.Module):
    def __init__(self, input_dim):
        super(LogisticRegression, self).__init__()
        # Додаємо лінійний шар: input_dim ознак → 1 вихід (бінарна класифікація)
        self.linear = nn.Linear(input_dim, 1)
        # Додаємо функцію активації сигмоїду
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        # Пряме проходження: дані йдуть від входу до виходу
        out = self.linear(x)    # z = wᵀX + b
        out = self.sigmoid(out) # ŷ = 1/(1+e^(-z)) → ймовірність [0, 1]
        return out              # Отримуємо передбачення
```

```python
# Створюємо екземпляр моделі
input_dim = X_train.shape[1]          # Вимірюємо кількість ознак: 18
model = LogisticRegression(input_dim)  # Отримуємо модель: Linear(18→1) + Sigmoid

# Визначаємо функцію втрат: Binary Cross-Entropy
criterion = nn.BCELoss()

# Визначаємо оптимізатор: SGD з learning rate = 0.01
optimizer = optim.SGD(model.parameters(), lr=0.01)
```

```python
# Переглядаємо структуру моделі та її параметри
model.parameters
```

```python
# Тренуємо модель: 50 епох (повних проходжень по даних)
num_epochs = 50
for epoch in range(num_epochs):
    # 1. Пряме проходження: подаємо дані → отримуємо передбачення
    outputs = model(X_train)
    # 2. Вимірюємо помилку (loss) між передбаченням і правильною відповіддю
    loss = criterion(outputs.squeeze(), y_train)
    
    # 3. Обнуляємо градієнти (інакше вони накопичуються!)
    optimizer.zero_grad()
    # 4. Обчислюємо градієнти — зворотне поширення (backpropagation)
    loss.backward()
    # 5. Оновлюємо ваги: w = w - lr × grad
    optimizer.step()
    
    # Виводимо loss кожні 5 епох
    if (epoch+1) % 5 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')
```
